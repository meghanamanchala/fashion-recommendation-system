# Fashion Recommendation System - EDA & Prototyping
This notebook demonstrates the exploratory data analysis (EDA) and content-based recommendation prototyping for the Myntra Fashion dataset.

```python
import os
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## 1. Load Dataset

```python
csv_path = os.path.join("..", "data", "styles.csv")
df = pd.read_csv(csv_path, on_bad_lines='skip')
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 2. Missing Values & Duplicate Checks

```python
print("Missing Values:")
print(df.isnull().sum())

print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

## 3. Exploratory Data Analysis

```python
# Gender Distribution
print("Gender Distribution:")
print(df['gender'].value_counts())

# Categories Distribution
print("\nMaster Categories:")
print(df['masterCategory'].value_counts().head(10))
```

## 4. Feature Engineering (Brand Extraction)

```python
def extract_brand(display_name):
    if pd.isna(display_name) or not isinstance(display_name, str):
        return "Unknown"
    tokens = display_name.strip().split()
    return tokens[0] if len(tokens) > 0 else "Unknown"

df['brand'] = df['productDisplayName'].apply(extract_brand)
print(f"Unique brands extracted: {df['brand'].nunique()}")
print(df['brand'].value_counts().head(10))
```

## 5. TF-IDF & Cosine Similarity Prototype

```python
# Combine metadata text
df['combined_features'] = (
    df['gender'].astype(str) + " " +
    df['articleType'].astype(str) + " " +
    df['baseColour'].astype(str) + " " +
    df['brand'].astype(str) + " " +
    df['usage'].astype(str) + " " +
    df['season'].astype(str)
)

# Apply TF-IDF
vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)
tfidf_matrix = vectorizer.fit_transform(df['combined_features'].head(1000)) # subset for prototyping

# Compute Cosine Similarity for the subset
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
print(f"Similarity Matrix Shape: {cosine_sim.shape}")
```
